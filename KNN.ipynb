# 📚 Theory

## 1. What is K-Nearest Neighbour?

**K-Nearest Neighbour (KNN)** is a **supervised, non-parametric,
instance-based machine learning algorithm**.

KNN classifies a new data point by examining the classes of its
**K nearest training samples**.

Unlike many machine learning algorithms, KNN does not build a complex
mathematical model during training. Instead, it stores the training
examples and performs distance calculations when a prediction is required.

### Basic Idea

> **Similar data points are likely to belong to the same class.**

For example, if a new flower is surrounded by flowers belonging mostly
to the `Iris-versicolor` class, KNN is likely to classify the new flower
as `Iris-versicolor`.

## 2. Working Principle of KNN

The KNN algorithm works using the following steps:

1. **Choose the value of K.**
2. Calculate the distance between the new sample and all training samples.
3. Select the **K nearest samples**.
4. Identify the classes of those neighbours.
5. Apply **majority voting**.
6. Assign the majority class to the new sample.

### Example

Suppose:

**K = 5**

The five nearest neighbours are:

| Class | Number of Neighbours |
|---|---:|
| Setosa | 1 |
| Versicolor | 3 |
| Virginica | 1 |

Since **Versicolor** has the majority vote:

**Predicted Class = Versicolor**

## 3. Distance Metrics

KNN uses a distance metric to determine how close two data points are.

### Euclidean Distance

The most commonly used distance metric is **Euclidean Distance**.

For two points:

$$
X=(x_1,x_2,...,x_n)
$$

and

$$
Y=(y_1,y_2,...,y_n)
$$

the Euclidean distance is:

$$
d(X,Y)=\sqrt{\sum_{i=1}^{n}(x_i-y_i)^2}
$$

For two-dimensional points:

$$
d=\sqrt{(x_1-y_1)^2+(x_2-y_2)^2}
$$

A **smaller distance** indicates that two samples are more similar.

---

### Manhattan Distance

Another distance metric is:

$$
d(X,Y)=\sum_{i=1}^{n}|x_i-y_i|
$$

It calculates the sum of absolute differences between corresponding features.

## 4. Importance of K

**K** represents the number of nearest neighbours considered when
making a prediction.

### Small K

Example:

**K = 1**

Advantages:

- Captures local patterns.
- Can model complex decision boundaries.

Disadvantages:

- Sensitive to noise.
- May cause **overfitting**.

### Large K

Advantages:

- More stable predictions.
- Less sensitive to individual noisy samples.

Disadvantages:

- May oversimplify the classification boundary.
- May cause **underfitting**.

Therefore, an appropriate value of **K** should be selected experimentally.

In this experiment, different K values are tested and compared.

## 5. Feature Scaling

Feature scaling is very important for KNN because KNN is a
**distance-based algorithm**.

Suppose one feature has values between:

**1 – 10**

while another feature has values between:

**1 – 10,000**

The second feature may dominate the distance calculation.

Therefore, we use **StandardScaler**.

### Standardization Formula

$$
z=\frac{x-\mu}{\sigma}
$$

Where:

- **x** = original value
- **μ** = mean of the feature
- **σ** = standard deviation

After standardization, the features have approximately:

- **Mean = 0**
- **Standard Deviation = 1**

### Important

The scaler is fitted only on the **training data** and then used to
transform the testing data.

This helps prevent **data leakage**.

## 6. Advantages and Limitations of KNN

### ✅ Advantages

- Simple and easy to understand.
- Easy to implement.
- No complex training process.
- Can be used for classification and regression.
- Works well with small datasets.
- Can model non-linear decision boundaries.

### ❌ Limitations

- Prediction can be computationally expensive for large datasets.
- Sensitive to feature scaling.
- Sensitive to noisy data.
- Choosing an appropriate K is important.
- Performance can decrease with high-dimensional data.

### 🔎 Explanation

The classification report provides four important measures:

- **Precision**

### Precision

$$
Precision=\frac{TP}{TP+FP}
$$

### Recall

$$

$$

### F1-Score

$$
F1=2\times

$$

**Support** represents the number of actual samples belonging to each class.

```python
# Cell 26: Generate Classification Report

print("KNN Classification Report:")

print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```

### CELL 27 — Test Different K Values

### 🔎 Explanation

This cell tests multiple K values from **1 to 15**.

For every K:

1. A KNN model is created.
2. The model is trained.
3. Predictions are generated.
4. Accuracy is calculated.

The results are stored in a DataFrame for comparison.

This experiment helps us understand the effect of K on model performance.

```python
# Cell 27: Evaluate Different K Values

k_values = range(1, 16)

k_accuracies = []

for k in k_values:

    model = KNeighborsClassifier(
        n_neighbors=k
    )

    model.fit(
        X_train_scaled,
        y_train
    )

    predictions = model.predict(
        X_test_scaled
    )

    score = accuracy_score(
        y_test,
        predictions
    )

    k_accuracies.append(score)

k_results = pd.DataFrame({
    "K": list(k_values),
    "Accuracy": k_accuracies
})

print(k_results)
```

### CELL 28 — K vs Accuracy Graph

### 🔎 Explanation

This graph represents the relationship between:

**K value → Classification Accuracy**

It helps us visually identify K values that provide good classification
performance.

The optimal K should be selected based on an appropriate evaluation
strategy rather than simply assuming that a particular K is always best.

```python
# Cell 28: Plot K Value versus Accuracy

plt.figure(figsize=(8, 5))

plt.plot(
    k_values,
    k_accuracies,
    marker="o"
)

plt.xlabel("Number of Neighbours (K)")
plt.ylabel("Accuracy")

plt.title("KNN Accuracy for Different K Values")

plt.xticks(
    list(k_values)
)

plt.grid(True)

plt.show()
```

### CELL 29 — Cross Validation

### 🔎 Explanation

This cell performs **5-fold cross-validation**.

The dataset is divided into five parts. The model is trained and
evaluated multiple times using different portions of the data.

The five accuracy values are displayed along with their mean.

Cross-validation provides a more reliable estimate of model performance
than relying only on one train-test split.

### Note

The K value shown here is selected from the K values tested earlier.
For a more rigorous model-selection workflow, K should ideally be selected
inside the cross-validation process itself.

```python
# Cell 29: Perform 5-Fold Cross-Validation

best_k = int(
    k_results.loc[
        k_results["Accuracy"].idxmax(),
        "K"
    ]
)

cv_model = KNeighborsClassifier(
    n_neighbors=best_k
)

scaler_cv = StandardScaler()

X_scaled_cv = scaler_cv.fit_transform(X)

cv_scores = cross_val_score(
    cv_model,
    X_scaled_cv,
    y,
    cv=5,
    scoring="accuracy"
)

print("Selected K:", best_k)

print("\nCross-Validation Scores:")
print(cv_scores)

print(
    "\nMean Cross-Validation Accuracy:",
    round(cv_scores.mean() * 100, 2),
    "%"
)
```

### CELL 30 — Final Prediction and Result

### 🔎 Explanation

This is the **final code cell** of the experiment.

It performs the following operations:

1. Creates a new unseen Iris flower sample.
2. Applies the same feature scaling used during model development.
3. Predicts the species using the KNN model.
4. Calculates **Precision, Recall and F1-score**.
5. Displays the initial K value.
6. Displays the best K tested.
7. Displays the cross-validation accuracy.
8. Displays the predicted species.

Thus, this cell provides the final experimental result.

```python
# Cell 30: Final Prediction and Result Summary

# Create a new unseen sample
new_sample = pd.DataFrame(
    [[5.9, 3.0, 5.1, 1.8]],
    columns=X.columns
)

# Scale the new sample
new_sample_scaled = scaler.transform(
    new_sample
)

# Predict the class
new_prediction = knn_model.predict(
    new_sample_scaled
)

# Calculate additional evaluation metrics
precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

print("=" * 60)
print("KNN FINAL RESULT")
print("=" * 60)

print(f"Initial K Value       : 5")
print(f"Test Accuracy         : {accuracy * 100:.2f}%")
print(f"Precision             : {precision:.4f}")
print(f"Recall                : {recall:.4f}")
print(f"F1-Score              : {f1:.4f}")
print(f"Best K Tested         : {best_k}")
print(f"Mean CV Accuracy      : {cv_scores.mean() * 100:.2f}%")

print("\nNew Sample:")
print(new_sample)

print(
    "\nPredicted Species:",
    new_prediction[0]
)

print("\nKNN experiment completed successfully.")
```
